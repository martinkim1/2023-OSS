```python
# 1. 정렬 알고리즘 함수를 직접 설계
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                # 두 숫자의 위치를 바꿈
    return arr

# 2. 데이터 준비 및 함수 실행
numbers = [3, 1, 4, 1, 5, 9, 2]
sorted_numbers = bubble_sort(numbers)

print(sorted_numbers) # [1, 1, 2, 3, 4, 5, 9]
```

```python
# 나이 계산 함수
def calculate_age(birth_year):
    current_year = 2026  # 현재 연도
    age = current_year - birth_year  # 나이 계산
    return age  # 나이 반환


```

```python
# 사용자 입력 받기
birth_year = int(input("태어난 연도를 입력하세요: "))

# 함수 호출하여 나이 계산
age = calculate_age(birth_year)
print('당신의 나이는', age, '세입니다.')
```

```python
class Car:
    def __init__(self, color):
        self.color = color
        self.speed = 0
    def accelerate(self, amount):
        self.speed += amount


car1 = Car("red") # 객체(인스턴스)
car2 = Car("blue") # 객체(인스턴스)
```

```python
# car1(빨간 차) 30만큼 가속하기
car1.accelerate(30)
```

```python
# 1. 계좌라는 객체를 만들기 위한 '설계도(클래스)' 정의
class Account:
    def __init__(self, owner, balance):
        self.owner = owner        # 속성 (예금주)
        self.balance = balance    # 속성 (잔액)

    def deposit(self, amount):    # 메서드 (입금 기능)
        self.balance += amount
        print(f"{amount}원이 입금되었습니다. 현재 잔액: {self.balance}원")

# 2. 설계도를 바탕으로 실제 메모리에 '철수의 계좌 객체'를 생성
chulsoo_account = Account("김철수", 10000)

# 3. 객체의 속성에 접근하거나 메서드를 실행
print(chulsoo_account.owner)  # 출력: 김철수 (속성 확인)
chulsoo_account.deposit(5000) # 출력: 5000원이 입금되었습니다. 현재 잔액: 15000원 (메서드 실행)
```
